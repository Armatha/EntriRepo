```python
# import required dependencies
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
```

```python
# create pandas dataframe from sklearn dataset
dataset=load_breast_cancer()
df=pd.DataFrame(dataset.data,columns=dataset.feature_names)
df['label']=dataset.target
# df.head()
```

```python
# to check imbalanced dataset
print(df['label'].value_counts())
print("yes, it is imbalance dataset")
# 0 : Malignant (cancerous)
# 1 : Benign (No cancer)
# solution: under sampling
Malignant=df[df['label']==0]
Benign=df[df['label']==1]
print(Malignant['label'].value_counts())
print(Benign['label'].value_counts())
```

```python
# under sampling the majority class which is 'Benign'
Benign_sampled=Benign.sample(len(Malignant))
print(Benign_sampled['label'].value_counts())
```

```python
Balanced_df=pd.concat([Malignant,Benign_sampled],axis=0)
Balanced_df.head()
print(f"value counts of balanced dataset is shown below:{Balanced_df['label'].value_counts()}")
```

```python
# input and target variable creation
X=df.drop(columns='label',axis=1)
y=df['label']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,stratify=y, random_state=2)
```

```python
# scaling
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
# LogisticRegression - used for binary classification & use sigmoid function to transform the output of a linear equation into a probability that ranges between 0 and 1. 
# DecisionTreeClassifier - single tree, split dataset into smaller subsets based on feature values, forming a tree-like structure
# RandomForestClassifier - Have (n_estimators) number of trees, each tree in forest works independently & final output is based on majority voting (classification) or averaging (regression).
# KNeighborsClassifier - classifies a data point based on the majority class of its k nearest neighbors, where k is number of neighbors considered for classification.
# SVC - works well with dataset having several numbers of columns and it is a small dataset (need high-dimensional spaces).
```

```python
# training & predicting then comparing models using evaluation metrics
models=[LogisticRegression(),DecisionTreeClassifier(),RandomForestClassifier(),KNeighborsClassifier(),SVC(kernel='linear')]
result={}
for model in models:
    model.fit(X_train_scaled,y_train)
    pred = model.predict(X_test_scaled)
    Accuracy=accuracy_score(y_test,pred)
    print(f"model name is {model.__class__.__name__}")
    print(f"its accuracy_score is {Accuracy}") 
    print("------------------------------------------------------")
    result[model.__class__.__name__]=Accuracy
    best=max(result,key=result.get)
    worst=min(result,key=result.get)
print(result)
print(f"the best model is {best}")
print(f"the worst model is {worst}")
```

```python
# better model performance features: High accuracy score
# By comparing the accuracy of each model, 
# the best model is LogisticRegression as it has high accuracy 
# the worst model is DecisionTreeClassifier as it has low accuracy.
```
