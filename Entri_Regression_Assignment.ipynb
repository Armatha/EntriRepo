```python
# import required dependencies
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
# create pandas dataframe from sklearn dataset
dataset=fetch_california_housing()
df=pd.DataFrame(dataset.data,columns=dataset.feature_names)
df['Price']=dataset.target
df.head()
```

```python
# null check
df.isnull().sum()
```

```python
# input and target variable creation
X=df.drop(columns='Price',axis=1)
y=df['Price']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2)
```

```python
# scaling
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
# LinearRegression - linear relationship between input features and target variable. y=wx+b
# DecisionTreeRegressor - single tree, split dataset into smaller subsets based on feature values, forming a tree-like structure
# RandomForestRegressor - Have (n_estimators) number of trees, each tree in forest works independently, meaning each one makes its own prediction
# XGBRegressor - trees are sequentially trained, where each tree tries to correct the errors of the previous tree
# SVR - Choose SVR if dataset has several numbers of columns and it is a small dataset (need high-dimensional spaces).
```

```python
# training & predicting then comapring models using evaluation metrics
models=[LinearRegression(),DecisionTreeRegressor(),RandomForestRegressor(),XGBRegressor(),SVR(kernel='linear')]
for model in models:
    model.fit(X_train_scaled,y_train)
    pred = model.predict(X_test_scaled)
    Mean_Absolute_Error=mean_absolute_error(y_test,pred)
    Mean_Squared_Error=mean_squared_error(y_test,pred)
    R2_Score=r2_score(y_test,pred)
    print(f"model name is {model}")
    print(f"its Mean_Absolute_Error is {Mean_Absolute_Error}")
    print(f"its Mean_Squared_Error is {Mean_Squared_Error}")
    print(f"its R2_Score is {R2_Score}")
    print("------------------------------------------------------")
```

```python
# better model performance features:
# 1. Low Mean_Squared_Error(MSE)
# 2. Low Mean_Absolute_Error (MAE)
# 3. High R2 score
# so by comaparing evaluation metrics,
# As XGBoost Regressor having low MSE, low MAE, high r2 score the best-performing algorithm: Gradient Boosting Regressor
# As SVR having high MSE, high MAE, low r2 score the worst-performing algorithm: Support Vector Regressor (SVR)
```
